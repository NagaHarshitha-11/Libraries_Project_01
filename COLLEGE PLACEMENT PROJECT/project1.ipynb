 # DATA ANALYSIS

```python
df.describe()
```

```python
print(df["Placement_Status"].value_counts())
```

```python
placement_counts = df["Placement_Status"].value_counts()

placed = placement_counts.get("Placed", 0)
total_students = len(df)

placement_percentage = (placed / total_students) * 100

print("Total Students:", total_students)
print("Placed Students:", placed)
print("Placement Percentage:", round(placement_percentage, 2), "%")
```

```python
pd.crosstab(
    df["Branch"],
    df["Placement_Status"]
)
```

```python
# Calculate branch-wise placement percentages 
branch_analysis = pd.crosstab(
    df["Branch"],
    df["Placement_Status"]
)

branch_analysis["Total Students"] = branch_analysis.sum(axis=1)

branch_analysis["Placement Percentage"] = (
    branch_analysis["Placed"]
    / branch_analysis["Total Students"]
    * 100
).round(2)

print(branch_analysis)
```

```python
known_status = df[df["Placement_Status"].isin(["Placed", "Not Placed"])]

branch_rate = (
    known_status.groupby("Branch")["Placement_Status"]
    .apply(lambda x: (x == "Placed").sum() / len(x) * 100)
    .round(2)
)

print(branch_rate.sort_values(ascending=False))
```

# **Visualizations**

```python
import matplotlib.pyplot as plt
```

# 1.Bar chart — Placement status

```python
branch_rate.plot(kind="bar")

plt.title("Branch-wise Placement Rate")
plt.xlabel("Branch")
plt.ylabel("Placement Percentage (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
cgpa_analysis = df[
    df["Placement_Status"].isin(["Placed", "Not Placed"])
].groupby("Placement_Status")["CGPA"].mean()

print(cgpa_analysis.round(2))
```

```python
cgpa_analysis.plot(
    kind="bar",
    figsize=(6, 4),
    rot=0
)

plt.title("Average CGPA by Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("Average CGPA")
plt.tight_layout()
plt.show()
```

## Conclusion

The College Placement Analysis project was developed to analyze student academic performance, technical skills, internship experience, and placement outcomes using Python and Pandas. The dataset was cleaned by handling missing values, correcting inconsistent categories, and standardizing data formats. Exploratory Data Analysis (EDA) and visualizations such as bar charts, pie charts, histograms, box plots, scatter plots, and correlation heatmaps were used to understand patterns in the data.

The analysis showed differences in placement rates across engineering branches. Mechanical Engineering (MECH) had the highest observed placement rate at 18.37%, while Civil Engineering (CIVIL) had the lowest at 9.41%, among students with known placement status. The project demonstrates how data analysis can help identify patterns between student characteristics and placement outcomes. These findings can support colleges in understanding placement trends and planning skill-development activities. However, the observed relationships do not necessarily prove that any single factor causes successful placement.
